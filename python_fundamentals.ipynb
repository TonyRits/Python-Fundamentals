## Basic Concepts

### 1. Variables

### 1.1 Understand the Concept

A variable is a name used to store a value in a Python program.

Variables help us store data and use that data later in the program.

Python does not require us to declare the data type separately.
The type is automatically determined from the value assigned to the variable.

### 1.2 Simple Example

A variable can be compared to a labeled box.

For example, if a box is labeled `name`, we can store a person's name inside it.
Later, we can use `name` to access that value.

```python
name = "Tony"
age = 23
course = "AI/ML"

print(name)
print(age)
print(course)
```

### 1.3 Explanation

In this program, I created three variables: `name`, `age`, and `course`.

The `name` variable stores a string value.
The `age` variable stores an integer value.
The `course` variable stores a string value.

The `print()` function is used to display the values stored in these variables.

## 3. Type Casting

### 3.1 Understand the Concept

Type casting means converting a value from one data type into another data type.

Python provides built-in functions such as `int()`, `float()`, `str()`, and `bool()` for converting values.

Type casting is useful when a program receives data in one type but needs to use it in another type.

### 3.2 Demonstrate the Concept

Suppose a user enters their age using `input()`.

The value received from `input()` is a string, even when the user enters a number.

For example, `"23"` is text, not an integer. If we want to perform mathematical operations with it, we can convert it into an integer using `int()`.

This demonstrates type casting because the value changes from one data type to another.

```python
age_text = "23"

print(age_text)
print(type(age_text))

age = int(age_text)

print(age)
print(type(age))
```

```python
price = 99

price_decimal = float(price)
price_text = str(price)

print(price_decimal)
print(type(price_decimal))

print(price_text)
print(type(price_text))
```

### 3.3 Explanation of Implementation

The program first stores `23` as a string in `age_text`.

The `int()` function converts the string `"23"` into the integer `23`.

The `float()` function converts an integer into a decimal number.

The `str()` function converts a value into a string.

The `type()` function is used to verify the data type before and after conversion.

## 4. Operators

### 4.1 Understand the Concept

Operators are symbols or keywords used to perform operations on values and variables.

Python provides different types of operators for mathematical calculations, comparisons, logical decisions, assigning values, checking membership, and checking object identity.

### 4.2 Demonstrate the Concept

Operators can be compared to the buttons on a calculator.

For example, the + operator adds two numbers, while the - operator subtracts one number from another.

Similarly, comparison and logical operators help a program make decisions based on conditions.

### Arithmetic Operators

```python
a = 20
b = 5

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Power:", a ** b)
```

### Comparison Operators

```python
a = 20
b = 10

print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
```

### Logical Operators

```python
age = 23
has_id = True

print(age >= 18 and has_id)
print(age >= 18 or has_id)
print(not has_id)
```

### Assignment Operators

```python
number = 10

number += 5
print(number)

number -= 2
print(number)

number *= 2
print(number)
```

### Membership Operators

```python
skills = ["Python", "SQL", "Machine Learning"]

print("Python" in skills)
print("Java" not in skills)
```

### Identity Operators

```python
a = [1, 2, 3]
b = a
c = [1, 2, 3]

print(a is b)
print(a is c)
print(a == c)
```

### 4.3 Explanation of Implementation

The arithmetic operators perform mathematical calculations.

The comparison operators compare two values and return either True or False.

The logical operators combine or reverse conditions.

The assignment operators modify or assign values to variables.

The membership operators check whether a value exists in a collection.

The identity operators check whether two variables refer to the same object.

In the last example, a and b refer to the same list, so a is b returns True. 
Although c contains the same values, it is a different list object, so a is c returns False.

## Control Flow

## 7. if / elif / else

### 7.1 Understand the Concept

Conditional statements allow a program to make decisions.

The if statement checks a condition first.

The elif statement checks another condition when the previous condition is false.

The else statement runs when none of the previous conditions are true.

### 7.2 Demonstrate the Concept

A student's grade can be determined based on their marks.

For example, if the marks are 90 or above, the student can receive an A grade.

If the marks are lower, another condition can be checked.

This demonstrates how a program makes different decisions based on a value.

```python
marks = 82

if marks >= 90:
    grade = "A"
elif marks >= 75:
    grade = "B"
elif marks >= 50:
    grade = "C"
else:
    grade = "Fail"

print("Marks:", marks)
print("Grade:", grade)
```

### 7.3 Explanation of Implementation

The program stores the student's marks in the marks variable.

The if statement checks whether the marks are 90 or above.

If that condition is false, the first elif condition is checked.

The next elif checks whether the marks are 50 or above.

If none of the conditions are satisfied, the else block assigns the grade as Fail.

This demonstrates how conditional statements allow a program to choose different actions based on conditions.

## 8. for Loop

### 8.1 Understand the Concept

A for loop is used to repeat a block of code for each item in a sequence or collection.

It is useful when we already know the items that we want to process or when we want to repeat an operation a specific number of times.

### 8.2 Demonstrate the Concept

Imagine checking the names of students one by one in a classroom.

Instead of writing separate code for every student, a for loop can process each name automatically.

This demonstrates iteration because the loop goes through each item one at a time.

```python
students = ["Tony", "Abhi", "Krish"]

for student in students:
    print(student)
```

```python
for number in range(1, 6):
    print(number)
```

### 8.3 Explanation of Implementation

The first program stores student names in a list.

The for loop takes each student from the list one at a time and prints the name.

In the second program, range() generates numbers from 1 to 5, and the for loop processes each number.

## 9. while Loop

### 9.1 Understand the Concept

A while loop repeatedly executes a block of code as long as a given condition remains True.

It is useful when the number of repetitions depends on a condition.

### 9.2 Demonstrate the Concept

Imagine a person continuing to practice badminton while there is still practice time available.

The activity continues while the condition is True.

This is similar to a while loop because the loop continues until its condition becomes False.

```python
count = 1

while count <= 5:
    print(count)
    count += 1
```

### 9.3 Explanation of Implementation

The variable count starts at 1.

The while loop checks whether count is less than or equal to 5.

After each iteration, count is increased by 1.

When count becomes 6, the condition becomes False and the loop stops.

## 10. break

### 10.1 Understand the Concept

The break statement is used to immediately stop a loop.

When Python reaches break, it exits the loop even if the loop condition could continue.

### 10.2 Demonstrate the Concept

Imagine searching for a particular student in a list.

Once the student is found, there is no need to continue searching.

The break statement can stop the loop at that point.

```python
students = ["Tony", "Abhi", "Krish", "Rahul"]

for student in students:
    print(student)

    if student == "Krish":
        break
```

### 10.3 Explanation of Implementation

The loop goes through the student names one by one.

When the value becomes "Krish", the if condition becomes True.

The break statement immediately stops the loop.

Therefore, the names after "Krish" are not processed.

## 11. continue

### 11.1 Understand the Concept

The continue statement skips the current iteration of a loop and moves to the next iteration.

Unlike break, continue does not stop the entire loop.

### 11.2 Demonstrate the Concept

Imagine checking a list of numbers but deciding to skip even numbers.

The loop can use continue to ignore the current number and move to the next one.

```python
for number in range(1, 6):

    if number % 2 == 0:
        continue

    print(number)
```

### 11.3 Explanation of Implementation

The loop checks numbers from 1 to 5.

When a number is even, the condition becomes True.

The continue statement skips that iteration.

Therefore, only the odd numbers are printed.

The loop itself continues running after each skipped iteration.

## 17. List Comprehension

### 1. Understand the Concept

List comprehension is a shorter way to create a new list.

It allows us to create a list by writing a loop and an expression in a single line.

## 23. Variable-Length Arguments

### 1. Understand the Concept

Variable-length arguments allow a function to accept a flexible number of arguments.

`*args` collects multiple positional arguments.

`**kwargs` collects multiple keyword arguments.

### 2. Demonstrate the Concept

Sometimes we do not know how many values will be provided to a function.

For example, a function calculating marks may receive marks for different numbers of subjects.

Variable-length arguments allow the function to handle this situation.

```python
def calculate_total(*marks):
    print("Total:", sum(marks))

calculate_total(80, 75, 90)
calculate_total(70, 85)
```

```python
def display_details(**details):
    for key, value in details.items():
        print(key, ":", value)

display_details(name="Tony", role="AI/ML Engineer", experience="Fresher")
```

### 3. Explanation of Implementation

In the first function, `*marks` collects all positional arguments into a tuple.

The `sum()` function calculates the total of all marks.

In the second function, `**details` collects keyword arguments into a dictionary.

The for loop accesses each key-value pair and displays it.

This demonstrates how functions can accept a flexible number of arguments.

## 25. Recursive Functions

### 1. Understand the Concept

A recursive function is a function that calls itself.

Recursion continues until a base condition is reached.

The base condition is important because it stops the recursive calls.

### 2. Demonstrate the Concept

Imagine counting down from a number.

The function can print the current number and then call itself with the next smaller number.

The process stops when the number reaches zero.

```python
def countdown(number):
    if number == 0:
        print("Finished")
        return

    print(number)
    countdown(number - 1)

countdown(5)
```

### 3. Explanation of Implementation

The function first checks the base condition `number == 0`.

If the condition is True, the function stops using `return`.

Otherwise, it prints the current number and calls itself with `number - 1`.

This process continues until the base condition is reached.

## 27. Generators

### 2. Demonstrate the Concept

The generator remembers where it stopped and continues from there.

```python
def generate_numbers():
    for number in range(1, 4):
        yield number

numbers = generate_numbers()

print(next(numbers))
print(next(numbers))
print(next(numbers))
```

### 3. Explanation of Implementation

The function uses `yield` to produce one value at a time.

## 29. Decorators

### 1. Understand the Concept

A decorator is a function that adds extra functionality to another function without changing its original code.

### 2. Demonstrate the Concept

Think of a decorator like adding an extra feature to an existing application.

The original function remains the same, but additional behavior is added around it.

```python
def logger(function):
    def wrapper():
        print("Function started")
        function()
        print("Function finished")
    return wrapper


@logger
def greet():
    print("Hello Tony")


greet()
```

### 3. Explanation of Implementation

`logger()` receives another function.

`wrapper()` adds messages before and after the original function.

The `@logger` syntax applies the decorator to `greet()`.

The original `greet()` function does not need to be modified.

## 30. Closures

### 1. Understand the Concept

A closure occurs when an inner function remembers and uses a variable from its outer function even after the outer function has finished executing.

### 2. Demonstrate the Concept

Imagine creating a personalized greeting function.

The outer function stores the person's name, and the inner function remembers that name.

```python
def create_greeting(name):
    def greet():
        print("Hello", name)
    return greet


greeting = create_greeting("Tony")
greeting()
```

### 3. Explanation of Implementation

`create_greeting()` receives a name.

The inner `greet()` function uses that name.

The outer function returns the inner function.

Even after `create_greeting()` finishes, the inner function remembers the name.

This is a closure.

## 31. global

### 1. Understand the Concept

The `global` keyword allows a function to modify a variable that was created outside the function at the global level.

### 2. Demonstrate the Concept

A global counter can be changed from inside a function by using the `global` keyword.

```python
count = 0

def increase_count():
    global count
    count += 1

increase_count()

print(count)
```

### 3. Explanation of Implementation

`count` is created outside the function.

The `global` keyword tells Python that the function should use the global `count`.

The function increases its value from 0 to 1.

## 32. nonlocal

### 1. Understand the Concept

The `nonlocal` keyword allows an inner function to modify a variable belonging to its nearest outer function.

### 2. Demonstrate the Concept

An outer function can contain a counter, and the inner function can update that counter using `nonlocal`.

```python
def counter():
    count = 0

    def increase():
        nonlocal count
        count += 1
        print(count)

    increase()
    increase()

counter()
```

### 3. Explanation of Implementation

`count` belongs to the outer `counter()` function.

The inner `increase()` function uses `nonlocal` to modify it.

The value changes from 0 to 1 and then to 2.

## 73. Singleton Pattern (Bonus)

### 1. Understand the Concept

The Singleton Pattern ensures that only one instance of a class is created.

It is useful when an application needs one shared object, such as a configuration manager.

### 2. Demonstrate the Concept

A configuration manager can be designed so that all parts of an application use the same instance.

```python
class Singleton:
    _instance = None

    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance


object1 = Singleton()
object2 = Singleton()

print(object1 is object2)
```

### 3. Explanation of Implementation

`_instance` stores the single object.

`__new__()` checks whether an object already exists.

If it does not exist, a new object is created.

If it already exists, the same object is returned.

Therefore, `object1` and `object2` refer to the same instance.

## 95. File Handlers

### 1. Understand the Concept

A FileHandler saves log messages into a file.
It is useful when logs need to be stored and checked later.

### 2. Demonstrate the Concept

```python
import logging

logger = logging.getLogger("app")
logger.setLevel(logging.INFO)

file_handler = logging.FileHandler("application.log")

formatter = logging.Formatter(
    "%(asctime)s - %(levelname)s - %(message)s"
)

file_handler.setFormatter(formatter)
logger.addHandler(file_handler)

logger.info("Application started")
logger.error("An error occurred")
```

## 97. Rotating Logs

### 1. Understand the Concept

Rotating logs automatically creates a new log file when the current file becomes large.
This prevents one log file from growing indefinitely.

### 2. Demonstrate the Concept

For example, we can limit a log file to a specific size and keep a few backup files.

```python
import logging
from logging.handlers import RotatingFileHandler

logger = logging.getLogger("rotating_app")
logger.setLevel(logging.INFO)

handler = RotatingFileHandler(
    "app.log",
    maxBytes=1000,
    backupCount=3
)

logger.addHandler(handler)

for i in range(20):
    logger.info(f"Log message {i}")
```

### 3. Explanation of Implementation

`RotatingFileHandler` creates a new file when the size limit is reached.
`backupCount=3` keeps up to three older log files.
